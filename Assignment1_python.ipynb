```python

string1 = "Hello "
name = input("Enter your Name: ")

string = string1 + name + ", welcome to Python programming"

print("First character:", string[0])

print("Last character:", string[-1])

print("First 5 characters:", string[:5])

print("Last 11 characters:", string[-11:])

print("Reverse string:", string[::-1])

start = string.index("Python")
print("Extracted word:", string[start:start+6])
```
